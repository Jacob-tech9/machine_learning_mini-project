```python
import pandas as pd
import numpy as np
```

```python
# Load datasets
primary_data = pd.read_csv('../data/raw/primary_data.csv', sep=";")
secondary_data = pd.read_csv('../data/raw/secondary_data.csv', sep=";")
display(primary_data.head())
display(secondary_data.head())

secondary_data.info()
```

```python
# We standardize and clean column names
primary_data.columns = (primary_data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","")
)

secondary_data.columns = (secondary_data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","")
)

primary_data.info()
secondary_data.info()
```

```python
# We decided not to merge the primary and secondary datasets as the secondary dataset does not include any unique identifiers. The primary dataset only includes 173 records.
```

```python
# Create a new dataframe of the 10 selected features

columns = [
    "class",
    "cap-color",
    "cap-shape",
    "does-bruise-or-bleed",
    "gill-color",
    "has-ring",
    "season",
    "habitat",
    "cap-diameter",
    "stem-height",
    "stem-width",
]

df = secondary_data[columns].copy()

# Check that the selected features have no missing values
df.info()
```
